# WWWMD 4

```python
# Import 10 Hz Data
wwwmd_4_10 = pd.read_csv(f'{cwd}/football_tracking_10hz.csv')
# Import 25 Hz Data
wwwmd_4_25 = pd.read_csv(f'{cwd}/football_tracking_25hz.csv')
```

```python
plt.plot(wwwmd_4_10['time'], wwwmd_4_10['position'], 'o-')
plt.show()
```

```python
plt.plot(wwwmd_4_25['time'], wwwmd_4_25['position'], 'o-')
plt.show()
```

# EXPLORING

```python
variance_10hz = wwwmd_4_10['position'].var()
variance_25hz = wwwmd_4_25['position'].var()

print(f"10 Hz variance: {variance_10hz}")
print(f"25 Hz variance: {variance_25hz}")
```

### Solution

In this case, the 25Hz data is not actually real data computed at that frame rate! This can be seen by plotting both sets of data on the same figure, and one can se that the 25Hz is simply the 10 Hz data but interpolated. This is a particularly egregious case where the data provider was not actually providing the product that their customer believed they were paying for, and is a key opportunity where skilled data analysts can provide value to organizations by spotting these inconsistencies.


```python

plt.plot(wwwmd_4_10['time'], wwwmd_4_10['position'],'o-', label = '10Hz')
plt.plot(wwwmd_4_25['time'], wwwmd_4_25['position'],'o-', label = '25Hz', alpha = 0.5)
plt.xlabel('Time')
plt.ylabel('Position')
plt.title('Frame Data')
plt.legend()
plt.show()
```
