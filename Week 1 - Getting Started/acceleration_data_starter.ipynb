# WWWMD 5

```python
# Restrict time to ~750
wwwmd_5 = pd.read_csv(f"{cwd}/acceleration_data.csv")
wwwmd_5.head()
```

```python
plt.plot(wwwmd_5['time'], wwwmd_5['acceleration'])
plt.xlabel('Time')
plt.ylabel('Acceleration')
plt.title('Acceleration Data')
plt.show()
```

# EXPLORING

```python
wwwmd_5['acceleration'].describe().T
```

```python
plt.figure(figsize=(8, 5))
plt.hist(wwwmd_5["acceleration"], bins=1000, edgecolor="black")
plt.xlabel("Acceleration")
plt.ylabel("Frequency")
plt.title("Histogram of Acceleration")
plt.show()
```

```python
below_zero = wwwmd_5[wwwmd_5["acceleration"] < 0]
share_below_zero = len(below_zero) / len(wwwmd_5)
print(f"Share of below zero: {share_below_zero*100:.1f}%")
```

```python
max_60 = wwwmd_5[wwwmd_5["acceleration"] == 60]
share_max_60 = len(max_60) / len(wwwmd_5)
print(f"Share of max 60: {share_max_60*100:.1f}%")
```

```python
# Histogram of Acceleration Data
plt.hist(wwwmd_5[wwwmd_5['acceleration']>5]['acceleration'], bins = 10)
plt.xlabel('Acceleration')
plt.ylabel('Count')
plt.xlim(0,100)
plt.title('Histogram of Acceleration Data')
plt.show()
```
